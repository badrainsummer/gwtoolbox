```python
import sources_new as sources, detectors_new  as detectors
```

```python
detector=detectors.earthbase.ligolike()
Sn=detector.NoiseCurve(flag=1)
```

```python
AntePattern=detector.AntePattern
```

```python
waveform=sources.kHz.BHB.waveform()
```

```python
rho_cri=8
Telfun=lambda z,m1,m2,chi: sources.kHz.BHB.TelFun(z, m1, m2, chi, rho_cri, waveform, Sn, AntePattern, flag=0)
CosMerRate=lambda z,m1,m2,chi: sources.kHz.BHB.CosMerRate(z,m1,m2,chi,model='fiducial')
```

```python
Ndet=lambda z, m1, m2, chi: sources.kHz.BHB.Ndet(20000,z,m1,m2,chi,Telfun,CosMerRate )
```

```python
Ndet(0.01,20,20,0.01)
```

```python
z_ranges=[1e-5,3]
m1_ranges=[5,50]
m2_ranges=m1_ranges
chi_ranges=[-0.5,0.5]
ranges=[z_ranges,m1_ranges,m2_ranges,chi_ranges]
```

```python
N_det=sources.kHz.BHB.TotNum(Ndet, ranges)
```

```python
print(int(N_det))
```

```python
init=[0.1, 20, 20, 0]
samples=sources.kHz.BHB.MCMCSample(Ndet, int(N_det), init)
```

```python
samples_z=samples[0]
samples_m1=samples[1]
samples_m2=samples[2]
samples_chi=samples[3]
print("z:", samples_z)
print("m1:", samples_m1) 
print("m2:", samples_m2)
print("chi:",samples_chi)
#print(samples_m1)
```

```python
from functions_new import *
from cosmology_new import cosmos
from astropy import units as u
from constants import *
import numpy as np
```

```python
errors_m1=np.zeros(len(samples[0]))
errors_m2=np.zeros(len(samples[0]))
errors_chi=np.zeros(len(samples[0]))
samples_DL=np.zeros(len(samples[0]))
errors_Mch=np.zeros(len(samples[0]))
samples_Mch=np.zeros(len(samples[0]))
errors_DL=np.zeros(len(samples[0]))
for i in range(len(samples[0])):
    z=samples_z[i]
    m1=samples_m1[i]
    m2=samples_m2[i]
    Mch=ChirpMass(m1,m2)
    samples_Mch[i]=Mch
    DL=cosmos.luminosity_distance(z).to_value(u.Gpc)
    A=CONST_AMPLITUDE_GW*Mch**(5./6.)/DL;
    chi=samples_chi[i]
    #print(A)
    PartMat=waveform.Partial(A,m1,m2,chi,0,0)
    f_up=waveform.freqLimit(m1,m2,chi)
    Fim=FisInfMatr(PartMat,Sn,f_up=f_up,numpar=5)
    ErrMt=Errors(Fim)
    #ErrMt_m1m2=ErrMt[np.ix_([0,1],[0,1])]
    Conv=FIMConvMatr(m1,m2,5)
    ErrMt_tilde=np.dot(np.dot(Conv.transpose(),ErrMt),Conv)
    errors_m1[i]=np.sqrt(np.diag(ErrMt))[0]
    errors_m2[i]=np.sqrt(np.diag(ErrMt))[1]
    errors_chi[i]=np.sqrt(np.diag(ErrMt))[2]
    samples_DL[i]=int(DL*1e3) # convert to Mpc
    #errors_Mch[i]=np.sqrt(np.diag(ErrMt_tilde))[0]
    #errors_DL[i]=errors_Mch[i]/Mch*5./6.*DL*1e3
```

```python
print("dm_1:",errors_m1)
print("dm_2:",errors_m2)
print("d_chi:",errors_chi)
print("DL(Mpc):",samples_DL)
#print("Mch:",samples_Mch)
#print("dMch:",errors_Mch)
#print("dDL(Mpc):",errors_DL)
```
